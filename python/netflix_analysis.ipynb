This beginner-friendly notebook follows the same functions as `pipeline.py`. It uses preprocessing, exploratory charts, KMeans clustering, and JSON export.

## 1. Import libraries and set paths

```python
from pathlib import Path
import sys
import pandas as pd

PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == 'python' else Path.cwd()
sys.path.append(str(PROJECT_ROOT / 'python'))
from pipeline import load_and_prepare, add_features, save_eda_plots, run_clustering, build_summary, write_outputs

INPUT_PATH = PROJECT_ROOT / 'data' / 'netflix_titles.csv'
OUTPUT_PATH = PROJECT_ROOT / 'server' / 'data'
```

## 2. Load and inspect the dataset

```python
# This function prints head, tail, info, describe, shape, and null counts before cleaning.
df = load_and_prepare(INPUT_PATH)
df.head()
```

## 3. Create genre flags, genre count, and scaled features

```python
# Adds the top-10 multi-hot genre columns and reports IQR/Z-score outliers.
df, feature_info = add_features(df)
feature_info['feature_columns']
```

## 4. Exploratory data analysis charts

```python
# Countplots, histograms, bars, heatmap, and boxplot are saved in python/plots.
save_eda_plots(df)
print('EDA charts saved in:', PROJECT_ROOT / 'python' / 'plots')
```

## 5. KMeans clustering

```python
# KMeans uses StandardScaler-scaled release year, duration, genre count, and genre flags.
cluster_info = run_clustering(df, feature_info)
df[['title', 'cluster']].head()
```

## 6. Cluster summary and interpretation

```python
summary = build_summary(df, cluster_info)
pd.DataFrame(summary['cluster_summary'])
```

## 7. Export files for the Express app

```python
write_outputs(df, summary, OUTPUT_PATH)
print('Chosen K:', summary['chosen_k'])
print('Silhouette score:', summary['silhouette_score'])
```
